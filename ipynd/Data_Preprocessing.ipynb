```python
# ==========================
# Import Required Libraries
# ==========================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries Imported Successfully!")
```

```python
# ==========================
# Load Datasets
# ==========================

train = pd.read_csv("train.csv")
features = pd.read_csv("features.csv")
stores = pd.read_csv("stores.csv")

print("Datasets Loaded Successfully!")
```

```python
# ==========================
# Data Types
# ==========================

print(train.dtypes)

print(features.dtypes)

print(stores.dtypes)
```

```python
# ==========================
# Dataset Information
# ==========================

print("TRAIN DATASET")
train.info()

print("\nFEATURES DATASET")
features.info()

print("\nSTORES DATASET")
stores.info()
```

```python
# ==========================
# Missing Values
# ==========================

print("TRAIN")
print(train.isnull().sum())

print("\nFEATURES")
print(features.isnull().sum())

print("\nSTORES")
print(stores.isnull().sum())
```

```python
# ==========================
# Missing Value Percentage
# ==========================

missing = features.isnull().sum()

missing_percent = (missing / len(features)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percent
})

missing_df.sort_values(by="Percentage", ascending=False)
```

```python
# ==========================
# Fill Missing Values
# ==========================

markdown_cols = [
    'MarkDown1',
    'MarkDown2',
    'MarkDown3',
    'MarkDown4',
    'MarkDown5'
]

features[markdown_cols] = features[markdown_cols].fillna(0)
features['CPI'] = features.groupby('Store')['CPI'].ffill()
features['Unemployment'] = features.groupby('Store')['Unemployment'].ffill()

features.isnull().sum()
```

```python
# ==========================
# Duplicate Check
# ==========================

print("Train Duplicates :", train.duplicated().sum())

print("Features Duplicates :", features.duplicated().sum())

print("Stores Duplicates :", stores.duplicated().sum())
```

```python
# ==========================
# Date Conversion
# ==========================

train['Date'] = pd.to_datetime(train['Date'])

features['Date'] = pd.to_datetime(features['Date'])

print(train.dtypes)
```

```python
# ==========================
# Merge Train and Features
# ==========================

merged_df = pd.merge(
    train,
    features,
    on=['Store', 'Date', 'IsHoliday'],
    how='left'
)

merged_df.head()
```

```python
# ==========================
# Merge Stores
# ==========================

merged_df = pd.merge(
    merged_df,
    stores,
    on='Store',
    how='left'
)

merged_df.head()
```

```python
# ==========================
# Final Dataset Shape
# ==========================

print(merged_df.shape)
```

```python
# ==========================
# Final Missing Values
# ==========================

merged_df.isnull().sum()
```

```python
# ==========================
# Date Features
# ==========================

merged_df['Year'] = merged_df['Date'].dt.year

merged_df['Month'] = merged_df['Date'].dt.month

merged_df['Week'] = merged_df['Date'].dt.isocalendar().week

merged_df['Quarter'] = merged_df['Date'].dt.quarter

merged_df['Day'] = merged_df['Date'].dt.day

merged_df.head()
```

```python
# ==========================
# Export Cleaned Dataset
# ==========================

merged_df.to_csv(
    "Walmart_Cleaned_Data.csv",
    index=False
)

print("Dataset Saved Successfully!")
```
